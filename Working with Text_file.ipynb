```python
##  Creating and writing to a text file

## Opn file in write mode

file = open('Employees.txt','w')

file.write('Abhay\n')
file.write('Rajan\n')
file.write('Aamir\n')
file.write('Sneha\n')

file.close()

print('File created successfully')
```

```python
## Reading entire file
with open('Employees.txt','r') as file:
  data = file.read()
print(data)
```

```python
# Reading file line by line

with open('Employees.txt', 'r') as file:
    for line in file:
        print(line.strip())
```

```python
## Using Readline

with open ('Employees.txt','r')as file:
  line1 = file.readline()
  line2 = file.readline()


print('First_line:',line1.strip())
print('Second_line:',line2.strip())
```

```python
## Using Readlines

with open('Employees.txt','r') as file:
  lines =file.readlines()
for line in lines:
  print(line.strip())
```

```python
## Appending Data as a file
with open('Employees.txt','a') as file:
  file.write('Karan\n')
  file.write('Akanksha\n')

print('New Student Added')
```

```python
## Writing user Input to a file
name = input("Enter the name :")
age = input('Enter the Age')

with open('Employees.txt','w')as file:
  file.write('Name:'+name+'\n')
  file.write('Age:'+age+'\n')

print('Information add')
```

```python
## Counting Number of Lines

with open('Employees.txt','r')as file:
  lines = file.readlines()


print('Number of words:',len(lines))
```

```python
## Counting words of text file

with open('Employees.txt','r')as file:
  data = file.read()

words = data.split()
print(len(words))
```

```python
## Counting Character

with open('Employees.txt','r')as file:
  data =  file.read()

print('Number of characters:',len(data))
```

```python
## To exclude space

character = data.replace(" ", '').replace('\n', '')
print("Character excluding spaces:",len(character))
```

```python
## Counting Occurence of a word

search_word = input('Enter a word:')

with open('Employees.txt','r')as file:
  data = file.read()

count = data.lower().count(search_word.lower())
print('Occurence:',count)
```

```python
## Copying Content from one file to another file

with open('Employees.txt','r') as file:
  data = file.read()

with open('Baackup.txt','w+') as file2:
  file2.write(data)
  file2.seek(0)
  print(file2.read())
```
